
## 4. Important NumPy Properties

| Property | Purpose |
|---|---|
| `.ndim` | Dimensions ki number |
| `.shape` | Har dimension ka size |
| `.size` | Total elements |
| `.dtype` | Elements ka data type |


```python
students = np.array([
    [80,85,95],
    [70,75,80],
    [90,95,100]
])

print(students.ndim)
print(students.shape)
print(students.size)
print(students.dtype)
```

```python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers[0])   # 10
print(numbers[2])   # 30
print(numbers[-1])  # 50
```

```python
print(numbers[1:4])
```

```python
students = np.array([
    [80, 85, 90],
    [70, 75, 80],
    [90, 95, 100]
])

print(students[0, 0])  # 80
print(students[1, 2])  # 80
print(students[2, 1])  # 95
```

```python
print(students[0, :])  # First student ke sab marks
print(students[:, 1])  # Sab students ke second subject ke marks
```

```python
print(students[0,0])
print(students[1,2])
print(students[2,:])
print(students[:,1])
```

```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])

print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.std(marks))
```

```python
print(np.mean(students, axis=0))
print(np.mean(students, axis=1))
```

```python
students = np.array([
    [80, 85, 90],
    [70, 75, 80],
    [90, 95, 100]
])

print("Total: ",np.sum(students))
print("Average: ", np.mean(students))
print("Highest: ",np.max(students))
print("Lowest: ",np.min(students))
print("Individual Average: ", np.mean(students,axis=1))
print("Subject Average: )", np.mean(students,axis=0))
```
