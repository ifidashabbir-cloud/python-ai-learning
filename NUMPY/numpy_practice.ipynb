NumPy mein hum arrays par mathematical calculations directly perform kar sakte hain.

**For example:**

- **Addition (`+`)**
- **Subtraction (`-`)**
- **Multiplication (`*`)**
- **Division (`/`)**

````python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([2, 4, 5, 8])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
````


## Reshaping Kya Hai?

Reshaping ka matlab array ke elements ko change kiye baghair uski shape change karna hai.

**Example:** Ek 1D array mein 6 elements hain. Hum usko 2D array mein convert kar sakte hain.

### 1D Array → 2D Array

**Original Shape:** `(6,)`

```python
[1, 2, 3, 4, 5, 6]
```

**New Shape:** `(2, 3)`

```python
[[1, 2, 3],
 [4, 5, 6]]
```


````python
import numpy as np
numbers = np.array([1,2,3,4,5,6])

new_array = numbers.reshape(2,3)

print("Reshape Array:",new_array)
print("New Array Shape:",new_array.shape)
````

````python
arr = np.arange(1,13)

new_array1 = arr.reshape(3,4)
new_array2 = arr.reshape(4,3)
new_array3 = arr.reshape(2,6)

print("New Array 01 Shape:",new_array1.shape)
print("New Array 02 Shape:",new_array2.shape)
print("New Array 03 Shape:",new_array3.shape)
````

````python
arr = np.arange(1,25)
reshape = arr.reshape(4,-1)
print(reshape)
print(reshape.shape)
````

````python
import numpy as np

marks = np.array([
    [60, 70, 80],
    [75, 85, 95],
    [50, 60, 70],
    [80, 90, 100]
])

print("Original Marks Shape:",marks.shape)
updated_marks = marks + 5
print("Updated Marks:",updated_marks)
print("Average Marks of every Student:",np.mean(updated_marks,axis=1))
print("Highest Marks:",updated_marks.max())
print("Lowest Marks:",updated_marks.min())
reshape = updated_marks.reshape(-1)
print("Flantted Array:",reshape)
````

# Day 10 - Advanced NumPy

## 1. Boolean Indexing

````python
numbers = np.array([12, 25, 8, 40, 17, 60, 5])

result = numbers[numbers > 20]

print(result)
````

````python
import numpy as np

ages = np.array([12,18,25,35,45,60])

result = ages[(ages >= 18) & (ages <= 40)]

print(result)
````

````python
marks = np.array([35, 50, 67, 82, 91, 45, 76, 100])

result = marks[(marks >= 50) & (marks <= 90)]

print(result)
````

np.where() Kya Hai?
np.where() NumPy ka function hai jo condition ke basis par values select ya replace karne ke kaam aata hai.

````python
import numpy as np

numbers = np.array([10, 25, 30, 45, 50])

result = np.where(numbers > 25)

print(result[0])
````

````python
marks = np.array([35, 60, 45, 80, 90])

result = np.where(marks >= 50, "Pass", "Fail")

print(result)
````

````python
scores = np.array([20, 55, 75, 40, 90, 30, 85])

result = np.where(scores >= 50, "Pass", "Fail")

print(result)
````

````python
import numpy as np

students_marks = np.random.randint(1,101,size=15)
print("All Marks: ",students_marks)

passing_marks = students_marks[students_marks >= 50]
failing_marks = students_marks[students_marks < 50]

result = np.where(students_marks >= 50, "Pass","Fail")
print("Results:",result)
total_pass = len(passing_marks)
print("Total Pass:",total_pass)
total_fail = len(failing_marks)
print("Total Fail:", total_fail)

average_marks = students_marks.mean()
print("Average Marks:", average_marks)

highest_marks = students_marks.max()
print("Highest Marks:", highest_marks)

lowest_marks = students_marks.min()
print("Lowest Marks:", lowest_marks)
````
